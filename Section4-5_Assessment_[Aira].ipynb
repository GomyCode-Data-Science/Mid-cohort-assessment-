The numerical analysis showed differences in the lengths of post titles and post bodies. The `body_length` variable showed more variation than the title length.

The correlation analysis was used to identify relationships between user IDs, post IDs, title length, and body length. Identifier variables such as `id` should be interpreted carefully because they do not represent meaningful continuous measurements.

Grouping posts by `userId` showed differences in average post body length between users.

An interesting pattern was that some posts had noticeably longer bodies than others. These records were investigated as potential unusual observations, but they were not removed because a long post is not necessarily an error.
