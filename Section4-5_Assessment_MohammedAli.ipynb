```python
# i have chosen the open meteo api to fetch data for nairobi kenya from 1st may 2026 to 24th august 2026.
url = "https://archive-api.open-meteo.com/v1/archive"
params = {
    "latitude":-1.2864 ,
    "longitude":36.8172,
    "start_date": "2026-05-01",
    "end_date": "2026-08-24",  
    "daily": "temperature_2m_max,temperature_2m_min,precipitation_sum,windspeed_10m_max",
    "timezone": "auto",
}
```

```python
# fetching 100 rows and saving to a pandas dataframe
response = requests.get(url, params=params)
data = response.json()['daily']
df = pd.DataFrame(data)
df
```

```python
# renaming the columns to more meaningful names 
df = df.rename(columns={
    "time": "date",
    "temperature_2m_max": "temp_max_c",
    "temperature_2m_min": "temp_min_c",
    "precipitation_sum": "precipitation_mm",
    "windspeed_10m_max": "windspeed_max_kmh",
})
```

```python
# printing the first 5 rows of the dataframe
print(f"Rows fetched: {len(df)}")
print(df.head())
```

```python
# saving the cleaned dataframe to a csv file
df.to_csv("nairobi_weather.csv", index=False)
print("Saved to nairobi_weather.csv")
```

```python
# Fixing the  data types (convert strings to dates/numbers as needed)
df['date'] = pd.to_datetime(df['date'])
df['temp_max_c'] = pd.to_numeric(df['temp_max_c'])
df['temp_min_c'] = pd.to_numeric(df['temp_min_c'])
df['precipitation_mm'] = pd.to_numeric(df['precipitation_mm'])
df['windspeed_max_kmh'] = pd.to_numeric(df['windspeed_max_kmh'])
print(df.dtypes)
df.head(10)
```

```python
# Remove or flag outliers (document your decision)
def flag_outliers(col):
    q1, q3 = df[col].quantile(0.25), df[col].quantile(0.75)
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    print(f"{col}: lower={lower}, upper={upper}")
    return (df[col] < lower) | (df[col] > upper)

for col in ["temp_max_c", "temp_min_c", "precipitation_mm", "windspeed_max_kmh"]:
    flag_col = f"{col}_outlier"
    df[flag_col] = flag_outliers(col)
    print(f"Outliers flagged in {col}: {df[flag_col].sum()}")
```

Part 3: Exploratory Data Analysis

```python
# reading the cleaned nairobi_weather csv file
df = pd.read_csv("nairobi_weather_cleaned.csv")
df["date"] = pd.to_datetime(df["date"])
```

```python
# creating a categorical column since the raw dataset is all numeric.'month' gives something meaningful to group weather patterns by.
df["month"] = df["date"].dt.month_name()
df
```

```python
# creating a list of numeric columns for further analysis
numeric_cols = ["temp_max_c", "temp_min_c", "precipitation_mm", "windspeed_max_kmh"]
print(numeric_cols)
```

```python
# calculating the mean, median, standard deviation, minimum, and maximum for each numeric column and rounding to 2 decimal places
summary = df[numeric_cols].agg(["mean", "median", "std", "min", "max"]).round(2)
print(summary)
```

```python
# finding value counts for 'month' 
print(df["month"].value_counts())
```

```python
#  finding mode for 'month'
print(df["month"].mode()[0])
```

```python
# calculating skewness for each numeric column and rounding to 2 decimal places
print(df[numeric_cols].skew().round(2))
```

```python
# calculating the correlation matrix for the numeric columns and rounding to 2 decimal places
print(df[numeric_cols].corr().round(2))
```

```python
#  calculating the average temp_max_c by month using groupby and rounding to 2 decimal places, then sorting in descending order
print(df.groupby("month")["temp_max_c"].mean().round(2).sort_values(ascending=False))
```

```python
# documenting my findings
with open("Section4-5_Findings_MohammedAli.txt", "a") as f:
    f.write("="*50 + "\n")
    f.write("\nSummary Statistics\n")
    f.write(summary.to_string())
    f.write("\n\n=== Categorical: Value Counts for 'month' ===\n")
    f.write(df["month"].value_counts().to_string())
    f.write("\n\n=== Categorical: Mode for 'month' ===\n")
    f.write(str(df["month"].mode()[0]))
    f.write("\n\n=== Skewness ===\n")
    f.write(df[numeric_cols].skew().round(2).to_string())
    f.write("\n\n=== Correlation Matrix (numeric pairs) ===\n")
    f.write(df[numeric_cols].corr().round(2).to_string())
    f.write("\n\n=== Groupby: Average temp_max_c by month ===\n")
    f.write(df.groupby("month")["temp_max_c"].mean().round(2).sort_values(ascending=False).to_string())
    f.write("\n" + "="*50 + "\n")
    
    
```
